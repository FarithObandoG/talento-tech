```python
import pandas as pd
# librerias
# Variables
m=0.849
b=1.509
Datos1 = [1,2,2,3,4,4,5,6]
Datos2 = [2,3,4,4,4,6,5,7]

#diccionario
diccionario = {'x':Datos1, 'y':Datos2}

#llamar pandas
df = pd.DataFrame(diccionario)
print(df)
#nuevas columnas
df['ypred'] = m*df['x']+b

df['y-pred'] = df['y']- df['ypred']

df ['y-pred^2'] = df['y-pred']**2
df
#fila con sumas
df.loc['Total'] = df.sum()

#mostrar dataframe
print(df)
df
```
